**02 - EDA**

 - top 10 cars and brands

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Main Title for the whole figure
fig.suptitle('Car Model Overview', fontsize=16, fontweight='bold')

# Graph 1: Top 10 Car Models (Clean Navy Blue)
sns.countplot(data=car_data, y='car_name', order=car_data['car_name'].value_counts().head(10).index, ax=axes[0], color='#1E3A8A')
axes[0].set_title('Top 10 Car Models')

# Graph 2: Top 10 Brands (Clean Deep Teal)
sns.countplot(data=car_data, y='brand', order=car_data['brand'].value_counts().head(10).index, ax=axes[1], color='#0F766E')
axes[1].set_title('Top 10 Car Brands')

plt.tight_layout()
plt.show()
```

```python
print("Average Age:", round(car_data['vehicle_age'].mean()))
print("Average KM Drivern :", round(car_data['km_driven'].mean()))
```

- seller information


```python

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Main Title for the whole figure
fig.suptitle('Seller Overview', fontsize=16, fontweight='bold')

# Left Chart: Seller Type Pie Chart
car_data['seller_type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['#1E3A8A', '#0F766E', '#64748B'],
    startangle=90,
    ylabel='',
    ax=axes[0]
)
axes[0].set_title('Seller Type Distribution')

# Right Chart: Average Selling Price by Seller Type
avg_price = round(car_data.groupby('seller_type')['selling_price'].mean())
avg_price.plot(
    kind='bar',
    color=['#1E3A8A', '#0F766E', '#64748B'],
    rot=0,  # Keeps x-axis labels horizontal
    ax=axes[1]
)
axes[1].set_title('Average Selling Price by Seller Type')
axes[1].set_xlabel('Seller Type')
axes[1].set_ylabel('Average Selling Price')

plt.tight_layout()
plt.show()
```

- car properties

```python

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Main Title for the whole figure
fig.suptitle('Car Properties Overview', fontsize=16, fontweight='bold')

# Chart 1: Fuel Type Pie Chart
car_data['fuel_type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['#1E3A8A', '#0F766E', '#64748B', '#94A3B8', '#CBD5E1'],
    startangle=90,
    ylabel='',
    ax=axes[0]
)
axes[0].set_title('Fuel Type')

# Chart 2: Transmission Type Bar Chart
car_data['transmission_type'].value_counts().plot(
    kind='bar',
    color=['#1E3A8A', '#0F766E'],
    rot=0,
    ax=axes[1]
)
axes[1].set_title('Transmission Type')
axes[1].set_xlabel('')
axes[1].set_ylabel('Count')

# Chart 3: Seats Bar Chart
car_data['seats'].value_counts().sort_index().plot(
    kind='bar',
    color='#0F766E',
    rot=0,
    ax=axes[2]
)
axes[2].set_title('Seating Capacity')
axes[2].set_xlabel('Seats')
axes[2].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
